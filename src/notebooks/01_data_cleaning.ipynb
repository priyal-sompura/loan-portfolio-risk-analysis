# **9. Null Value Imputation**

```python
loan_type_df['borrower_cycle_type'] = loan_type_df['borrower_cycle_type'].fillna(
    loan_type_df['borrower_cycle_type'].mode()[0]
)

loan_type_df['min_number_of_repayments'] = loan_type_df['min_number_of_repayments'].fillna(
    loan_type_df['min_number_of_repayments'].median()
)

loan_type_df['max_number_of_repayments'] = loan_type_df['max_number_of_repayments'].fillna(
    loan_type_df['max_number_of_repayments'].median()
)
```

**9.1. Customer and Group Table Missing Value Treatment**

```python
# Missing staff IDs are replaced with the most frequent value because staff assignment is a categorical attribute.
customer_df['staff_id'] = customer_df['staff_id'].fillna(
    customer_df['staff_id'].mode()[0]
)

# Filling missing staff IDs using the mode to preserve the most common assignment pattern.
group_df['staff_id'] = group_df['staff_id'].fillna(
    group_df['staff_id'].mode()[0]
)
```

**9.2 Loan Table Missing Value Treatment**

```python
# Missing loan counters are replaced with 0,
# assuming no previous loans exist for the record.
loan_df['loan_counter'] = loan_df['loan_counter'].fillna(0)

# Missing product counters are replaced with 0,
# indicating no prior product usage history.
loan_df['loan_product_counter'] = loan_df['loan_product_counter'].fillna(0)

# Missing loan purpose category values are filled
# using the most frequently occurring category.
loan_df['loanpurpose_cv_id'] = loan_df['loanpurpose_cv_id'].fillna(
    loan_df['loanpurpose_cv_id'].mode()[0]
)

# Missing loan purpose IDs are filled with the mode
# to maintain consistency in loan categorization.
loan_df['loan_purpose_id'] = loan_df['loan_purpose_id'].fillna(
    loan_df['loan_purpose_id'].mode()[0]
)
```

**9.3 Loan Type Table Missing Value Treatment**

```python

# Missing arrears tolerance amounts are replaced with 0,assuming no tolerance amount was configured.
loan_type_df['arrearstolerance_amount'] = loan_type_df['arrearstolerance_amount'].fillna(0)

# Missing instalment multiples are replaced with 0,indicating no specific instalment multiple restriction.
loan_type_df['instalment_amount_in_multiples_of'] = loan_type_df['instalment_amount_in_multiples_of'].fillna(0)

# Missing overdue day values are filled with the mode to maintain consistency in NPA classification rules.
loan_type_df['overdue_days_for_npa'] = loan_type_df['overdue_days_for_npa'].fillna(
    loan_type_df['overdue_days_for_npa'].mode()[0]
)

# Missing fund IDs are replaced using the most frequent value to minimize impact on product-level analysis.
loan_type_df['fund_id'] = loan_type_df['fund_id'].fillna(
    loan_type_df['fund_id'].mode()[0]
)

# Missing borrower cycle types are filled using the mode because this is a categorical loan product attribute.
loan_type_df['borrower_cycle_type'] = loan_type_df['borrower_cycle_type'].fillna(
    loan_type_df['borrower_cycle_type'].mode()[0]
)

# Missing minimum repayment counts are replaced with the median to reduce the influence of extreme values.
loan_type_df['min_number_of_repayments'] = loan_type_df['min_number_of_repayments'].fillna(
    loan_type_df['min_number_of_repayments'].median()
)

# Missing maximum repayment counts are replaced with the median because the median is robust against skewed distributions.
loan_type_df['max_number_of_repayments'] = loan_type_df['max_number_of_repayments'].fillna(
    loan_type_df['max_number_of_repayments'].median()
)
```
